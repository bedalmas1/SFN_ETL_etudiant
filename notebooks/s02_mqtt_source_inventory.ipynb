# Séquence 2 — Inventorier une source MQTT
L'instantané observé couvre-t-il les cinq zones attendues ?

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
from iot_decision.source_inventory import load_jsonl, inventory, load_expected, diagnose
```

## TP 1 — Observer puis TP 2 — Inventorier

```python
records = load_jsonl(ROOT / 'data/samples/batch002_retained_messages.jsonl')
rows = inventory(records)
print('messages observés :', len(rows))
for row in rows: print(row['zone'], row['topic'])
```

## TP 3 — Comparer à l'attendu

```python
expected = load_expected(ROOT / 'data/samples/batch002_expected_sensors.csv')
report = diagnose(rows, expected)
print(report)
assert report.observed_count == 4 and report.expected_count == 5
assert report.complete is False
```
