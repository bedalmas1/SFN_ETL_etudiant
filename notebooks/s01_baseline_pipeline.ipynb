## Observer sans modifier
Relevez effectif, zones, deux horloges et retained. Que peut-on décider?

```python
records=load_raw(sample)
len(records),records[0].keys(),records[0]['payload'].keys()
```

## Structurer
Qu'est-ce qui devient plus lisible sans devenir plus vrai?

```python
rows=transform_raw(records)
zones=sorted({r['zone'] for r in rows})
[(z,sum(r['zone']==z for r in rows)) for z in zones]
```

## Tester une question
Calculez le maximum par zone, comparez à 35 °C et dites ce que le résumé masque.

```python
maxima={z:max(r['value'] for r in rows if r['zone']==z) for z in zones}
maxima
```

## Contradicteur
Comparez measured_at à 10:05Z. Une zone fragilise-t-elle la confiance? Ne supprimez rien.

```python
[(r['zone'],r['measured_at']) for r in rows]
```

## Votre recommandation
Rédigez décision, confiance, deux preuves, deux incertitudes et vérification. L'appel suivant teste seulement l'API.

```python
brief=recommend(rows)
assert brief.confidence in {'très faible','faible','moyen','élevé'}
```
